# Home Credit Default Risk - Exploratory Data Analysis (EDA)
This notebook performs thorough data profiling, missing value analysis, demographic distributions, and financial metric correlations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.loader import load_and_join_datasets, initialize_database

initialize_database()
df = load_and_join_datasets()
print('Dataset Shape:', df.shape)
```

## 1. Target Distribution & Class Imbalance

```python
target_dist = df['TARGET'].value_counts(normalize=True) * 100
print(f'Default Rate: {target_dist.get(1, 0):.2f}%')
df['TARGET'].value_counts().plot(kind='bar', color=['#28a745', '#dc3545'])
plt.title('Target Distribution (0 = Non-Default, 1 = Default)')
plt.ylabel('Count')
plt.show()
```

## 2. Demographic & Income Analysis

```python
inc_summary = df.groupby('NAME_INCOME_TYPE')['TARGET'].agg(['count', 'mean'])
inc_summary['default_rate_%'] = inc_summary['mean'] * 100
print(inc_summary)
```
